```python
import os
from ebooklib import epub
import pymupdf
import re
import shutil
```

```python

# Directory containing the PDF files
pdf_dir = r'C:\Users\blain\Calibre Library\Core Knowledge'

final_dir = "converted_epubs"

# Iterate through the PDF files in the directory
for root, dirs, files in os.walk(pdf_dir):
    

    for filename in files:

        if 'Core Knowledge.original_epub' in filename:
            print("Skipping Original EPUB file")
            continue

        if 'Core Knowledge' in filename:
            book = epub.read_epub(os.path.join(root, filename))
            title = book.title
            source = os.path.join(root, filename)
            destination = os.path.join(final_dir, title + '.epub')
            # shutil.copy(source, destination)
            print(f"Moved to: {destination}")
```
